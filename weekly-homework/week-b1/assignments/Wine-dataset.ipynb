### Correlation matrix

A correlation matrix showing the linear correlation between each feature
and the target feature ‘Quality’. This will give you the correlation coefficient
between each of the features and the target feature. Values close to +1 or -
1 are strongly related while values close to 0 are not related.

```python
plt.figure(figsize = (20, 10))
sns.heatmap(wine_df.corr(), annot = True) # Annotate each cell with the correlation value
```

A correlation matrix is a table showing correlation coefficients between variables. Each cell in the table shows the correlation between two variables.
